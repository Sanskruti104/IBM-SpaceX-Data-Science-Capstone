# 2. Web Scraping — Falcon 9 Launch Records
This notebook reads the Wikipedia launch-history tables using pandas' HTML table reader. Check the page structure and source terms before reuse; scraped values should be validated against API records.

```python
import pandas as pd
from pathlib import Path

ROOT = Path("..")
RAW = ROOT / "data" / "raw"
RAW.mkdir(parents=True, exist_ok=True)

url = "https://en.wikipedia.org/wiki/List_of_Falcon_9_and_Falcon_Heavy_launches"
tables = pd.read_html(url)
print("Tables found:", len(tables))
[(i, t.shape, list(t.columns)[:5]) for i, t in enumerate(tables[:10])]
```

```python
# Inspect all tables and select the launch-history table after reviewing the preview.
for i, table in enumerate(tables):
    print(f"TABLE {i}: shape={table.shape}")
    display(table.head(3))
```

```python
# Save a source snapshot for reproducibility; choose the table index after inspecting previews.
# The Wikipedia page can change. Update TABLE_INDEX if the page layout changes.
TABLE_INDEX = next((i for i, t in enumerate(tables) if t.shape[0] > 20 and t.shape[1] >= 4), None)
if TABLE_INDEX is None:
    raise ValueError("Could not automatically identify a launch table. Inspect the previews and set TABLE_INDEX manually.")
scraped = tables[TABLE_INDEX].copy()
scraped.to_csv(RAW / "wikipedia_launch_table_raw.csv", index=False)
print("Saved table index:", TABLE_INDEX, "shape:", scraped.shape)
scraped.head()
```
